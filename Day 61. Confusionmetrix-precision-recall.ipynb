```python
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv('/kaggle/input/datasets/fedesoriano/heart-failure-prediction/heart.csv')
```

```python
df.head()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
x = df.iloc[:,0:11]
y = df.iloc[:,-1]
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

```python
# Creating Pipeline

cat_cols =['Sex','ChestPainType','RestingECG','ExerciseAngina','ST_Slope']
```

```python
cat_pipe = Pipeline([
    ('OnehotEncoder',OneHotEncoder(drop='first'))
])
```

```python
preprocessing = ColumnTransformer([
    ('cat_pipe',cat_pipe,cat_cols)
],remainder = 'passthrough')
```

```python
pipe = Pipeline([
    ('preprocess',preprocessing),
    ('model',LogisticRegression())
])
```

```python
pipe.fit(x_train,y_train)
```

```python
y_pred = pipe.predict(x_test)
print('Prediction Accuracy :',accuracy_score(y_test,y_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
pipe1= Pipeline([
    ('preprocess',preprocessing),
    ('model1',DecisionTreeClassifier())
])
```

```python
pipe1.fit(x_train,y_train)
```

```python
y_pred_new = pipe1.predict(x_test)
print("Decision Tree accuracy",accuracy_score(y_test,y_pred_new))
```

```python
from sklearn.metrics import confusion_matrix,precision_score,recall_score,f1_score
```

```python
print('Logistic Regression')
cm = pd.DataFrame(confusion_matrix(y_test,y_pred),columns=list(range(0,2)))
print(cm)

print('precision score ',precision_score(y_test,y_pred))
print('recall score ', recall_score(y_test,y_pred))
print('f1 Score ',f1_score(y_test,y_pred))
```

```python
print('Decision Tree Classifier Regression')
cm = pd.DataFrame(confusion_matrix(y_test,y_pred_new),columns=list(range(0,2)))
print(cm)

print('precision score ',precision_score(y_test,y_pred_new))
print('recall score ', recall_score(y_test,y_pred_new))
print('f1 Score ',f1_score(y_test,y_pred_new))
```
