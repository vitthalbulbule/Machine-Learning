# Underfitting
## Model not work better because the data has not linear 

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Data
X = np.linspace(-3, 3, 100).reshape(-1, 1)
y = X.ravel()**2 + np.random.normal(0, 1, 100)

# Model
model = LinearRegression()
model.fit(X, y)

# Prediction
y_pred = model.predict(X)

# Plot
plt.scatter(X, y)
plt.plot(X, y_pred, linewidth=3,color='red')
plt.show()
```

# Overfitting
## Model tries to learn each and every thing , also including noise and many more from the data

```python
from sklearn.preprocessing import PolynomialFeatures

# Small dataset
X = np.array([1,2,3,4,5,6,7,8]).reshape(-1,1)
y = np.array([1,4,9,16,25,36,49,64])

# High-degree polynomial
poly = PolynomialFeatures(degree=10)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)

# Prediction
x_new = np.linspace(1,8,100).reshape(-1,1)
x_new_poly = poly.transform(x_new)
y_pred = model.predict(x_new_poly)

# Plot
plt.scatter(X, y)
plt.plot(x_new, y_pred, linewidth=3,color='red')
plt.show()
```
